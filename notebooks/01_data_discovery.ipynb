```python
csv_files = sorted(RAW_DATA_DIR.glob("*.csv"))

print(f"Number of CSV files found: {len(csv_files)}")

for file_path in csv_files:
    print(file_path.name)
```

```python
file_inventory = []

for file_path in csv_files:
    file_inventory.append(
        {
            "file_name": file_path.name,
            "file_size_mb": round(file_path.stat().st_size / (1024 ** 2), 2),
        }
    )

file_inventory_df = pd.DataFrame(file_inventory)

file_inventory_df
```

```python
dataset_files = {
    "customers": RAW_DATA_DIR / "olist_customers_dataset.csv",
    "geolocation": RAW_DATA_DIR / "olist_geolocation_dataset.csv",
    "order_items": RAW_DATA_DIR / "olist_order_items_dataset.csv",
    "payments": RAW_DATA_DIR / "olist_order_payments_dataset.csv",
    "reviews": RAW_DATA_DIR / "olist_order_reviews_dataset.csv",
    "orders": RAW_DATA_DIR / "olist_orders_dataset.csv",
    "products": RAW_DATA_DIR / "olist_products_dataset.csv",
    "sellers": RAW_DATA_DIR / "olist_sellers_dataset.csv",
    "category_translation": RAW_DATA_DIR / "product_category_name_translation.csv",
}

dataset_files
```

```python
for dataset_name, file_path in dataset_files.items():
    print(
        f"{dataset_name:<22} "
        f"exists={file_path.exists()} "
        f"file={file_path.name}"
    )
```

## Looking at samples of each dataset

```python
sample_data = {}

for dataset_name, file_path in dataset_files.items():
    try:
        sample_df = pd.read_csv(file_path, nrows=5)
        sample_data[dataset_name] = sample_df

        print(
            f"{dataset_name:<22} "
            f"loaded successfully — "
            f"{sample_df.shape[0]} rows, "
            f"{sample_df.shape[1]} columns"
        )

    except Exception as error:
        print(f"{dataset_name:<22} failed: {error}")
```

```python
sample_data["customers"]
```

```python
for dataset_name, sample_df in sample_data.items():
    print(f"\n{dataset_name.upper()}")
    print("-" * 60)

    for column in sample_df.columns:
        print(column)
```

## basic schema summary

```python
schema_summary = []

for dataset_name, file_path in dataset_files.items():
    sample_df = pd.read_csv(file_path, nrows=100)

    for column_name, dtype in sample_df.dtypes.items():
        schema_summary.append(
            {
                "dataset": dataset_name,
                "column_name": column_name,
                "sample_inferred_dtype": str(dtype),
            }
        )

schema_summary_df = pd.DataFrame(schema_summary)

schema_summary_df
```

# Load complete datasets in dictionary of dataframes

```python
# Loading all datasets in a dictionary of data frames
datasets = {}

for dataset_name, file_path in dataset_files.items():
    try:
        df = pd.read_csv(file_path)

        datasets[dataset_name] = df

        print(
            f"{dataset_name:<22} "
            f"{df.shape[0]:>10,} rows × "
            f"{df.shape[1]:>2} columns"
        )

    except Exception as error:
        print(f"{dataset_name:<22} failed: {error}")
```

- The table contains 1,000,163 rows and 5 columns.
- One row represents one recorded latitude/longitude location for a ZIP-code prefix.

- Latitude and longitude contain some extreme values outside Brazil’s normal geographic range, so coordinate outliers should be investigated during cleaning.
- For analytics, the geolocation table may later be aggregated to one representative latitude and longitude per ZIP-code prefix.
